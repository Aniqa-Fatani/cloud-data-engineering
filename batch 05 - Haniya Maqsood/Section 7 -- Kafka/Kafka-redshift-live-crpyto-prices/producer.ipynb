```python
# pip install kafka-python requests
import json, time, requests
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers=['16.171.10.142:9092'],
    key_serializer=lambda k: k.encode('utf-8'),
    value_serializer=lambda v: json.dumps(v).encode('utf-8'),
)
SYMBOLS = ['BTCUSDT', 'ETHUSDT', 'SOLUSDT', 'BNBUSDT']
URL = 'https://api.binance.com/api/v3/ticker/price'

while True:
    resp = requests.get(URL, params={'symbols': json.dumps(SYMBOLS, separators=(',', ':'))}, timeout=5)
    resp.raise_for_status()
    now = time.strftime('%Y-%m-%d %H:%M:%S', time.gmtime())
    for item in resp.json():
        event = {'symbol': item['symbol'], 'price': float(item['price']), 'ts': now}
        print('event', event)
        producer.send('crypto_price', key=item['symbol'], value=event)
    time.sleep(2)
```
