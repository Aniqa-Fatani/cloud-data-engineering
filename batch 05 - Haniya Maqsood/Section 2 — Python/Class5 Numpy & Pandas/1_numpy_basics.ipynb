## 1.1 — What is NumPy and Why Do We Need It?

**NumPy** = **Num**erical **Py**thon

Imagine you want to find the average goals scored across **852 World Cup matches**.
With a normal Python list you'd write a loop — slow and messy.
With NumPy: `np.mean(goals)` — done in one word, **instantly**.

| Normal Python List | NumPy Array |
|--------------------|-------------|
| Slow with large data | **Very fast** |
| No built-in math | **100s of math functions** |
| Manual loops needed | **Operations on whole array at once** |

Today we'll use NumPy to analyse **real World Cup match data** — goals, attendance, scores.

```python
# Always import NumPy first — everyone uses 'np' as the short name
import numpy as np
import pandas as pd    # we'll need this just to load the CSV

print("NumPy version:", np.__version__)
print("Ready! ✔")
```

## 1.2 — Loading the Data & Extracting NumPy Arrays

```python
# Load the CSV with Pandas (we'll learn Pandas properly in Hour 2)
# For now, think of this as just 'opening the file'
raw = pd.read_csv("WorldCupMatches.csv")

# Clean up: remove rows with missing values
raw = raw.dropna(subset=["Year","Home Team Name","Home Team Goals","Away Team Goals"])

# Convert to correct types
raw["Year"]            = raw["Year"].astype(int)
raw["Home Team Goals"] = raw["Home Team Goals"].astype(int)
raw["Away Team Goals"] = raw["Away Team Goals"].astype(int)
raw["Attendance"]      = raw["Attendance"].astype(float)

print(f"✔ Dataset loaded — {len(raw)} World Cup matches from {raw['Year'].min()} to {raw['Year'].max()}")
print(f"Columns available: {raw.columns.tolist()}")
```

```python
# ── Extract data into NumPy arrays ──
# Think of each array as one column pulled out of the table

home_goals  = np.array(raw["Home Team Goals"])   # goals scored by home team
away_goals  = np.array(raw["Away Team Goals"])   # goals scored by away team
attendance  = np.array(raw["Attendance"])         # crowd size at each match
years       = np.array(raw["Year"])               # year of each match

print("Home goals array (first 10):", home_goals[:10])
print("Away goals array (first 10):", away_goals[:10])
print("Array type:", type(home_goals))
print("Shape:", home_goals.shape)    # (852,) → 852 matches, 1 dimension
```

## 1.3 — Array Math (Broadcasting)

With NumPy, you can do **math on the whole array at once** — no loop needed!
This is called **broadcasting** — the operation spreads across every element.

```python
# ── Total goals in each match ──
# Adding two arrays adds them ELEMENT BY ELEMENT (match by match)
total_goals = home_goals + away_goals
print("home goals", home_goals)
print(away_goals,"away goals")
print("Total goals per match (first 10):", total_goals[:10])
print()

# ── Home advantage check ──
# True/False array — was the home team winning?
home_winning = home_goals > away_goals
home_drawing = home_goals == away_goals
away_winning = away_goals > home_goals

print(f"Matches where HOME team won : {np.sum(home_winning)}")
print(f"Matches that were DRAWS     : {np.sum(home_drawing)}")
print(f"Matches where AWAY team won : {np.sum(away_winning)}")
print()

# ── Home win percentage ──
home_win_pct = (np.sum(home_winning) / len(home_winning)) * 100
print(f"Home win rate: {home_win_pct:.1f}%  ← does playing at home help?")
```

## 1.4 — Statistical Functions on Real Match Data

```python
# ── Goals statistics ──
print("=" * 45)
print("  WORLD CUP GOALS ANALYSIS (1930–2014)")
print("=" * 45)
print(f"  Total matches played  : {len(total_goals)}")
print(f"  Total goals scored    : {np.sum(total_goals)}")
print(f"  Average goals/match   : {np.mean(total_goals):.2f}")
print(f"  Median goals/match    : {np.median(total_goals):.1f}")
print(f"  Highest scoring match : {np.max(total_goals)} goals")
print(f"  Lowest scoring match  : {np.min(total_goals)} goals")
print(f"  Std deviation         : {np.std(total_goals):.2f}  (spread of scores)")
print()

# Find WHICH match had the most goals
highest_idx = np.argmax(total_goals)   # index of the highest value
print(highest_idx, "highest index")
match = raw.iloc[highest_idx]
print(match.head(25),"match")
print(f"  Most goals in one match: {match['Year']} — "
      f"{match['Home Team Name']} {int(match['Home Team Goals'])} - "
      f"{int(match['Away Team Goals'])} {match['Away Team Name']}")
```

## 1.5 — Indexing & Slicing (Getting What You Need)

You can grab specific matches or ranges using index numbers — just like a list, but faster.

```python
# ── Slicing — get specific matches ──
print("First 5 matches ever played at the World Cup:")
print(total_goals[:5])
print()

print("Last 5 matches in the dataset:")
print(total_goals[-5:])
print()

# ── Boolean Filtering — find high-scoring matches ──
# 'Give me total_goals, but only the ones where goals > 6'
thrillers = total_goals[total_goals > 6]
print(f"Matches with MORE than 6 goals: {len(thrillers)}")
print(f"Their scores: {sorted(thrillers, reverse=True)}")
print()

# ── Filter by YEAR ──
# Get goals only from 2014 World Cup
goals_2014 = total_goals[years == 2014]
print(f"2014 World Cup — {len(goals_2014)} matches")
print(f"Average goals per match in 2014: {np.mean(goals_2014):.2f}")
```

## 1.6 — Grouping with NumPy — Goals by World Cup Year

```python
# ── Average goals per match for each World Cup year ──
unique_years = np.unique(years)     # sorted list of unique years

print(f"{'Year':<8} {'Matches':>8} {'Total Goals':>12} {'Avg Goals/Match':>16}")
print("-" * 50)

for yr in unique_years:
    mask        = years == yr              # True for matches in this year
    yr_goals    = total_goals[mask]        # only goals from this year
    num_matches = len(yr_goals)
    total_yr    = np.sum(yr_goals)
    avg_yr      = np.mean(yr_goals)
    print(f"{yr:<8} {num_matches:>8} {total_yr:>12} {avg_yr:>16.2f}")
```

Use NumPy to compare World Cup goals across different decades.

```python
# 🎯 ACTIVITY 1 — Which DECADE was the most exciting (most goals)?

decades = {
    "1930s": (1930, 1939),
    "1950s": (1950, 1959),
    "1960s": (1960, 1969),
    "1970s": (1970, 1979),
    "1980s": (1980, 1989),
    "1990s": (1990, 1999),
    "2000s": (2000, 2009),
    "2010s": (2010, 2019),
}

print(f"{'Decade':<10} {'Matches':>8} {'Total Goals':>12} {'Avg/Match':>12}")
print("=" * 46)

best_decade = ""
best_avg    = 0

for decade, (start, end) in decades.items():
    mask     = (years >= start) & (years <= end)
    # print(mask,"mask")
    d_goals  = total_goals[mask]
    if len(d_goals) == 0:
        continue
    d_total  = np.sum(d_goals)
    d_avg    = np.mean(d_goals)
    print(f"{decade:<10} {len(d_goals):>8} {d_total:>12} {d_avg:>12.2f}")
    if d_avg > best_avg:
        best_avg    = d_avg
        best_decade = decade

print("=" * 46)
print(f"\n🏆 Most exciting decade: {best_decade} ({best_avg:.2f} goals/match)")
```

```python
# 🔥 Bonus: Find the percentage of matches in the dataset that were 0-0 draws
goalless = np.sum((home_goals == 0) & (away_goals == 0))
print(f"0-0 Draws : {goalless} matches ({goalless/len(total_goals)*100:.1f}% of all games)")

# 🔥 Bonus 2: Find matches where the losing team scored 3 or more goals
# (Those are exciting even in defeat!)
big_losses = total_goals[(total_goals >= 6) & (np.minimum(home_goals, away_goals) >= 3)]
print(f"High-scoring close matches (6+ goals, loser scored 3+): {len(big_losses)}")
```
